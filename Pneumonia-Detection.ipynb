#**Chest X-ray Image Classification for Pneumonia Detection Using CNN and Transfer Learning**

**Models Implemented:**

1. Custom CNN Model:

 * 4 Conv2D layers (32, 64, 128, 256 filters) with BatchNormalization, MaxPooling, and Dropout.

 * Final Dense layers with sigmoid activation for binary classification.

2. Pre-trained Models (Transfer Learning):

 * DenseNet121, Xception, MobileNetV2 (trained and evaluated).

3. Vision Transformer (ViT):

 * Attempted using vit-keras but commented out due to hardware limitations/errors (not executed in the final code).

**Training Setup:**
* Image preprocessing: Resizing (128x128), augmentation (rotation, shifts, zoom, flip).

* Class imbalance: Visualized but not explicitly addressed (e.g., no class weighting).

* Callbacks: ModelCheckpoint, EarlyStopping, ReduceLROnPlateau.

**Evaluation:**

 * Metrics: Accuracy, AUC, confusion matrix, classification report.









```python
!pip install vit-keras
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import classification_report, confusion_matrix
import tensorflow as tf
from tensorflow.keras.models import Sequential, Model
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout, BatchNormalization, GlobalAveragePooling2D
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import ModelCheckpoint, EarlyStopping, ReduceLROnPlateau
from tensorflow.keras.applications import (
    VGG16, VGG19, ResNet50, ResNet101, DenseNet121, DenseNet201,
    MobileNetV2, EfficientNetB0, Xception, InceptionV3, InceptionResNetV2
)
from tensorflow.keras.applications.convnext import ConvNeXtSmall
from tensorflow.keras.applications import MobileNetV3Small
# from vit_keras import vit

# Set random seed for reproducibility
tf.random.set_seed(42)
np.random.seed(42)

# Constants
IMG_SIZE = (128, 128)
BATCH_SIZE = 16
EPOCHS = 30
NUM_CLASSES = 1  # Binary classification (pneumonia vs normal)
INPUT_SHAPE = IMG_SIZE + (3,)

from google.colab import files
from google.colab import files
import zipfile
import os

# 1. Upload the ZIP file
print("Please upload your chest_xray.zip file")
uploaded = files.upload()

# 2. Extract to the RIGHT location
for filename in uploaded.keys():
    if filename.endswith('.zip'):
        print("Extracting to /content/chest_xray...")
        with zipfile.ZipFile(filename, 'r') as zip_ref:
            zip_ref.extractall('/content/chest_xray')  # Note: content/chest_xray (no extra folders)
        print("Done!")

# 3. Verify
print("\nExtracted files:")
!ls -la /content/chest_xray



base_dir = '/content/chest_xray/chest_xray'

print("Train path exists?", os.path.exists(os.path.join(base_dir, 'train')))
print("Test path exists?", os.path.exists(os.path.join(base_dir, 'test')))


train_dir = os.path.join(base_dir, 'train')
test_dir = os.path.join(base_dir, 'test')

import os
print("Current working directory:", os.getcwd())
# Create validation set from training data (since original dataset doesn't have val folder)
train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=20,
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
    fill_mode='nearest',
    validation_split=0.2  # Using 20% of training data for validation
)

test_datagen = ImageDataGenerator(rescale=1./255)


# Load data
train_generator = train_datagen.flow_from_directory(
    train_dir,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='binary',
    color_mode='rgb',
    shuffle=True,
    subset='training'  # This is for training data
)

val_generator = train_datagen.flow_from_directory(
    train_dir,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='binary',
    color_mode='rgb',
    shuffle=False,
    subset='validation'  # This is for validation data
)

test_generator = test_datagen.flow_from_directory(
    test_dir,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='binary',
    color_mode='rgb',
    shuffle=False
)

# Class distribution visualization
def plot_class_distribution(generator, title):
    class_counts = generator.classes
    unique, counts = np.unique(class_counts, return_counts=True)
    plt.figure(figsize=(8, 6))
    sns.barplot(x=['Normal', 'Pneumonia'], y=counts)
    plt.title(title)
    plt.ylabel('Number of Images')
    plt.show()

plot_class_distribution(train_generator, 'Train Set Class Distribution')
plot_class_distribution(val_generator, 'Validation Set Class Distribution')
plot_class_distribution(test_generator, 'Test Set Class Distribution')

# Sample images visualization
def plot_sample_images(generator, num_images=8):
    images, labels = next(generator)
    plt.figure(figsize=(15, 10))
    for i in range(num_images):
        plt.subplot(4, 4, i+1)
        plt.imshow(images[i])
        plt.title('Pneumonia' if labels[i] == 1 else 'Normal')
        plt.axis('off')
    plt.tight_layout()
    plt.show()

plot_sample_images(train_generator)

# Custom CNN Model
def build_custom_cnn(input_shape=INPUT_SHAPE):
    model = Sequential([
        Conv2D(32, (3, 3), activation='relu', input_shape=input_shape),
        BatchNormalization(),
        MaxPooling2D((2, 2)),

        Conv2D(64, (3, 3), activation='relu'),
        BatchNormalization(),
        MaxPooling2D((2, 2)),

        Conv2D(128, (3, 3), activation='relu'),
        BatchNormalization(),
        MaxPooling2D((2, 2)),

        Conv2D(256, (3, 3), activation='relu'),
        BatchNormalization(),
        MaxPooling2D((2, 2)),

        Flatten(),
        Dense(512, activation='relu'),
        Dropout(0.5),
        Dense(1, activation='sigmoid')
    ])

    model.compile(optimizer=Adam(learning_rate=0.0001),
                  loss='binary_crossentropy',
                  metrics=['accuracy', tf.keras.metrics.AUC(name='auc')])

    return model

# Function to create transfer learning model
def build_transfer_model(base_model, input_shape=INPUT_SHAPE, freeze_layers=True):
    if freeze_layers:
        for layer in base_model.layers:
            layer.trainable = False

    x = base_model.output
    x = GlobalAveragePooling2D()(x)
    x = Dense(256, activation='relu')(x)
    x = Dropout(0.5)(x)
    predictions = Dense(1, activation='sigmoid')(x)

    model = Model(inputs=base_model.input, outputs=predictions)

    model.compile(optimizer=Adam(learning_rate=0.0001),
                  loss='binary_crossentropy',
                  metrics=['accuracy', tf.keras.metrics.AUC(name='auc')])

    return model

# Callbacks
def get_callbacks(model_name):
    checkpoint = ModelCheckpoint(
        f'best_{model_name}.h5',
        monitor='val_accuracy',
        save_best_only=True,
        mode='max',
        verbose=1
    )

    early_stopping = EarlyStopping(
        monitor='val_loss',
        patience=5,
        restore_best_weights=True,
        verbose=1
    )

    reduce_lr = ReduceLROnPlateau(
        monitor='val_loss',
        factor=0.2,
        patience=3,
        min_lr=1e-7,
        verbose=1
    )

    return [checkpoint, early_stopping, reduce_lr]

# Function to train and evaluate model
def train_and_evaluate(model, model_name, train_gen, val_gen, test_gen, epochs=EPOCHS):
    print(f"\nTraining {model_name}...")
    history = model.fit(
        train_gen,
        epochs=epochs,
        validation_data=val_gen,
        callbacks=get_callbacks(model_name),
        verbose=1
    )

    # Plot training history
    plt.figure(figsize=(12, 5))

    plt.subplot(1, 2, 1)
    plt.plot(history.history['accuracy'], label='Train Accuracy')
    plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
    plt.title('Accuracy over epochs')
    plt.legend()

    plt.subplot(1, 2, 2)
    plt.plot(history.history['loss'], label='Train Loss')
    plt.plot(history.history['val_loss'], label='Validation Loss')
    plt.title('Loss over epochs')
    plt.legend()

    plt.tight_layout()
    plt.show()

    # Evaluate on test set
    print("\nEvaluating on test set...")
    test_loss, test_acc, test_auc = model.evaluate(test_gen, verbose=1)
    print(f"Test Accuracy: {test_acc:.4f}")
    print(f"Test AUC: {test_auc:.4f}")

    # Classification report and confusion matrix
    y_pred = model.predict(test_gen)
    y_pred = (y_pred > 0.5).astype(int)
    y_true = test_gen.classes

    print("\nClassification Report:")
    print(classification_report(y_true, y_pred, target_names=['Normal', 'Pneumonia']))

    plt.figure(figsize=(6, 6))
    cm = confusion_matrix(y_true, y_pred)
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=['Normal', 'Pneumonia'],
                yticklabels=['Normal', 'Pneumonia'])
    plt.title('Confusion Matrix')
    plt.ylabel('True Label')
    plt.xlabel('Predicted Label')
    plt.show()

    return history

# Train Custom CNN
custom_cnn = build_custom_cnn()
custom_cnn.summary()
custom_cnn_history = train_and_evaluate(custom_cnn, 'custom_cnn', train_generator, val_generator, test_generator)

# Dictionary of pre-trained models to evaluate
pretrained_models = {
    'VGG16': VGG16(weights='imagenet', include_top=False, input_shape=INPUT_SHAPE),
    'VGG19': VGG19(weights='imagenet', include_top=False, input_shape=INPUT_SHAPE),
    'ResNet50': ResNet50(weights='imagenet', include_top=False, input_shape=INPUT_SHAPE),
    'ResNet101': ResNet101(weights='imagenet', include_top=False, input_shape=INPUT_SHAPE),
    'DenseNet121': DenseNet121(weights='imagenet', include_top=False, input_shape=INPUT_SHAPE),
    'DenseNet201': DenseNet201(weights='imagenet', include_top=False, input_shape=INPUT_SHAPE),
    'MobileNetV2': MobileNetV2(weights='imagenet', include_top=False, input_shape=INPUT_SHAPE),
    'MobileNetV3': MobileNetV3Small(weights='imagenet', include_top=False, input_shape=INPUT_SHAPE),
    'EfficientNetB0': EfficientNetB0(weights='imagenet', include_top=False, input_shape=INPUT_SHAPE),
    'Xception': Xception(weights='imagenet', include_top=False, input_shape=INPUT_SHAPE),
    'InceptionV3': InceptionV3(weights='imagenet', include_top=False, input_shape=INPUT_SHAPE),
    'InceptionResNetV2': InceptionResNetV2(weights='imagenet', include_top=False, input_shape=INPUT_SHAPE),
    'ConvNeXt': ConvNeXtSmall(weights='imagenet', include_top=False, input_shape=INPUT_SHAPE),
}

# Train and evaluate selected pre-trained models
selected_models = ['DenseNet121', 'Xception', 'MobileNetV2']

model_histories = {}
for model_name in selected_models:
    base_model = pretrained_models[model_name]
    model = build_transfer_model(base_model)
    model.summary()
    history = train_and_evaluate(model, model_name, train_generator, val_generator, test_generator)
    model_histories[model_name] = history

# # Vision Transformer (ViT) implementation
# def build_vit_model(input_shape=INPUT_SHAPE):
#     vit_model = vit.vit_b16(
#         image_size=input_shape[0],
#         activation='sigmoid',
#         pretrained=True,
#         include_top=False,
#         pretrained_top=False,
#         classes=1
#     )

#     model = Sequential([
#         vit_model,
#         Flatten(),
#         Dense(256, activation='relu'),
#         Dropout(0.5),
#         Dense(1, activation='sigmoid')
#     ])

#     model.compile(optimizer=Adam(learning_rate=0.0001),
#                   loss='binary_crossentropy',
#                   metrics=['accuracy', tf.keras.metrics.AUC(name='auc')])

#     return model

# Train ViT model if you have sufficient resources
# try:
#     vit_model = build_vit_model()
#     vit_model.summary()
#     vit_history = train_and_evaluate(vit_model, 'vit', train_generator, val_generator, test_generator, epochs=15)
#     model_histories['ViT'] = vit_history
# except Exception as e:
#     print(f"Could not train ViT model due to: {str(e)}")
#     print("This might be due to hardware limitations. Try with smaller image size or batch size.")

# Compare all models' performance
def compare_models(histories, test_generator):
    results = []

    for model_name, history in histories.items():
        # Load the best saved model
        try:
            model = tf.keras.models.load_model(f'best_{model_name.lower()}.h5')
        except:
            print(f"Could not load saved model for {model_name}, using last trained weights")
            model = history.model

        # Evaluate
        test_loss, test_acc, test_auc = model.evaluate(test_generator, verbose=0)
        results.append({
            'Model': model_name,
            'Test Accuracy': test_acc,
            'Test AUC': test_auc,
            'Val Accuracy': max(history.history['val_accuracy']),
            'Train Accuracy': max(history.history['accuracy'])
        })

    results_df = pd.DataFrame(results)
    results_df = results_df.sort_values('Test Accuracy', ascending=False)

    plt.figure(figsize=(12, 6))
    sns.barplot(data=results_df, x='Model', y='Test Accuracy')
    plt.title('Model Comparison - Test Accuracy')
    plt.xticks(rotation=45)
    plt.show()

    return results_df

# Compare models if we have more than one
if len(model_histories) > 1:
    comparison_results = compare_models(model_histories, test_generator)
    print(comparison_results)



# Save and download ONLY the best model (MobileNetV2)
best_model = model_histories['MobileNetV2'].model

# Save in modern .keras format (recommended)
best_model.save('pneumonia_model.keras')

# Download the model
from google.colab import files
# files.download('pneumonia_model.h5') #HDF format
files.download('pneumonia_model.keras') #Native keras format
```
